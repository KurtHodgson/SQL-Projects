## Let's explore Steam Data from around 2019 and see what factors play into the success rate of games. 

```python
import duckdb as db
```

```python
steam = db.read_csv("Data/extended_steam_games.csv").df()
sql=db.sql
steam
```

```python
##need to find the midpoint of owners so we can actually use it as a reference.
sql("""
SELECT 
    format(' {:,.0f}', (
        replace(string_split(owners, '..')[1], ',', '')::BIGINT + 
        replace(string_split(owners, '..')[2], ',', '')::BIGINT
    ) / 2) AS avg_owners  
FROM steam
group by avg_owners
""")
```
