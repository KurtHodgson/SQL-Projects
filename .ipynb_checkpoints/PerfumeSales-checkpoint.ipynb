```python
mens = db.read_csv("Data/ebay_mens_perfume.csv").df()
womens= db.read_csv("Data/ebay_womens_perfume.csv").df()
sql = db.sql
mens
```

```python
#lets join these 2 tables and use them together.
sql(""" COPY(
SELECT *
FROM mens
UNION ALL
SELECT *
FROM womens)
TO 'menandwomenperfume.csv' (HEADER, DELIMITER ',');
""")
```

```python
#lets now add that new table to what we're working with so we don't need to keep unioning.
perfume = db.read_csv("Data/menandwomenperfume.csv").df()
```

```python
sql("""
SELECT 
    brand,
    SUM(sold) AS total_sold
FROM perfume
WHERE brand IS NOT NULL
GROUP BY brand
ORDER BY total_sold DESC
LIMIT 20;
""").df()
```

```python
#since there are 2 Giorgio Beverly Hills, lets try making everything lowercase for the group by.
sql("""
SELECT 
    LOWER(brand) AS brand,
    SUM(sold) AS total_sold
FROM perfume
WHERE brand IS NOT NULL
GROUP BY LOWER(brand)
ORDER BY total_sold DESC
LIMIT 20;
""").df()
```
